```python
import re

def clean_text(text):
    text = str(text)

    # lowercase
    text = text.lower()

    # remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # remove user mentions
    text = re.sub(r"@\w+", "", text)

    # remove hashtag symbol but keep hashtag word
    text = re.sub(r"#", "", text)

    # remove unnecessary special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["clean_comment"] = df["comment"].apply(clean_text)

df[["comment", "clean_comment"]].head(10)
```

```python
df["calculated_comment_length"] = (
    df["comment"]
    .astype(str)
    .str.len()
)
```

```python
df["calculated_word_count"] = (
    df["comment"]
    .astype(str)
    .str.split()
    .str.len()
)
```

```python
df["calculated_engagement"] = (
    df["comment_likes"]
    + df["comment_retweets"]
    + df["comment_replies"]
)
```

```python
topic_performance = (
    df.groupby("topic")
    .agg(
        total_posts=("post_id", "count"),
        total_likes=("comment_likes", "sum"),
        total_retweets=("comment_retweets", "sum"),
        total_replies=("comment_replies", "sum"),
        total_engagement=("total_engagement", "sum"),
        avg_engagement=("total_engagement", "mean"),
        avg_sentiment=("sentiment_score", "mean")
    )
    .round(2)
    .sort_values(
        "total_engagement",
        ascending=False
    )
)

topic_performance
```
